```python

from pathlib import Path
import pandas as pd
from rdkit import Chem, rdBase

print("RDKit version:", rdBase.rdkitVersion)

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

RAW_DIR = PROJECT_ROOT / "data" / "raw"
RESULTS_DIR = PROJECT_ROOT / "results" / "metrics"
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

validation_rows = []
invalid_rows = []

for file in sorted(RAW_DIR.glob("*.csv")):
    df = pd.read_csv(file)

    for idx, smiles in df["smiles_standarized"].items():
        mol = Chem.MolFromSmiles(str(smiles)) if pd.notna(smiles) else None

        validation_rows.append({
            "file": file.name,
            "row_index": idx,
            "valid_smiles": mol is not None
        })

        if mol is None:
            invalid_rows.append({
                "file": file.name,
                "row_index": idx,
                "smiles": smiles,
                "label": df.loc[idx, "label"]
            })

validation = pd.DataFrame(validation_rows)
invalid = pd.DataFrame(
    invalid_rows,
    columns=["file", "row_index", "smiles", "label"]
)

print("Total structures checked:", len(validation))
print("Valid structures:", int(validation["valid_smiles"].sum()))
print("Invalid structures:", int((~validation["valid_smiles"]).sum()))

print("\nValidity by file:")
display(
    validation.groupby("file")["valid_smiles"]
    .agg(["count", "sum"])
    .rename(columns={"count": "total", "sum": "valid"})
)

print("\nExamples of invalid structures:")
display(invalid.head(20))

validation.to_csv(
    RESULTS_DIR / "smiles_validation.csv", index=False
)
invalid.to_csv(
    RESULTS_DIR / "invalid_smiles_review.csv", index=False
)

print("\nAudit files saved to:", RESULTS_DIR)
```

```python

from pathlib import Path
import pandas as pd
from rdkit import Chem

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
RESULTS_DIR = PROJECT_ROOT / "results" / "metrics"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

valid_records = []
invalid_records = []

for file in sorted(RAW_DIR.glob("*.csv")):
    df = pd.read_csv(file)

    isoform, split = file.stem.rsplit("_", 1)

    for row_idx, row in df.iterrows():
        smiles = row["smiles_standarized"]
        mol = Chem.MolFromSmiles(str(smiles)) if pd.notna(smiles) else None

        record = {
            "source_file": file.name,
            "source_row_index": row_idx,
            "isoform": isoform.upper(),
            "split": split,
            "smiles_original": smiles,
            "label": row["label"],
            "group": row["group"]
        }

        if mol is None:
            invalid_records.append(record)
            continue

        record["smiles_canonical"] = Chem.MolToSmiles(
            mol, canonical=True
        )
        valid_records.append(record)

cleaned = pd.DataFrame(valid_records)
invalid = pd.DataFrame(invalid_records)

# Save the cleaned modeling dataset and the invalid records separately.
cleaned.to_csv(
    PROCESSED_DIR / "cyp_substrates_cleaned.csv",
    index=False
)
invalid.to_csv(
    RESULTS_DIR / "invalid_structures_for_review.csv",
    index=False
)

print("Original records:", len(valid_records) + len(invalid_records))
print("Valid records retained:", len(cleaned))
print("Invalid records set aside:", len(invalid))
print("Missing canonical SMILES:", cleaned["smiles_canonical"].isna().sum())

print("\nValid records per isoform and split:")
display(
    cleaned.groupby(["isoform", "split"])
    .size()
    .unstack(fill_value=0)
)

print("\nInvalid records — full SMILES:")
if len(invalid):
    with pd.option_context("display.max_colwidth", None):
        display(invalid[[
            "source_file", "source_row_index",
            "smiles_original", "label"
        ]])
else:
    print("No invalid records.")

print("\nSaved cleaned dataset to:")
print(PROCESSED_DIR / "cyp_substrates_cleaned.csv")
```

```python

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

CLEANED_FILE = (
    PROJECT_ROOT / "data" / "processed"
    / "cyp_substrates_cleaned.csv"
)
RESULTS_DIR = PROJECT_ROOT / "results" / "metrics"
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(CLEANED_FILE)

# How many distinct compounds occur in each partition?
print("Distinct canonical compounds by isoform and partition:")
display(
    df.groupby(["isoform", "split"])["smiles_canonical"]
      .nunique()
      .unstack(fill_value=0)
)

# For each compound, record every partition in which it appears.
compound_partitions = (
    df.groupby("smiles_canonical")["split"]
      .agg(lambda values: sorted(set(values)))
      .reset_index(name="partitions")
)

compound_partitions["number_of_partitions"] = (
    compound_partitions["partitions"].str.len()
)

cross_partition_compounds = compound_partitions[
    compound_partitions["number_of_partitions"] > 1
].copy()

print(
    "\nDistinct compounds appearing in multiple partitions:",
    len(cross_partition_compounds)
)

display(cross_partition_compounds.head(20))

# Count compounds that appear in each pair of partitions.
partition_sets = {
    split: set(df.loc[df["split"] == split, "smiles_canonical"])
    for split in ["train", "val", "test"]
}

pairwise_overlap = []
for a, b in [
    ("train", "val"),
    ("train", "test"),
    ("val", "test")
]:
    shared = partition_sets[a] & partition_sets[b]
    pairwise_overlap.append({
        "partition_1": a,
        "partition_2": b,
        "shared_canonical_compounds": len(shared)
    })

print("\nGlobal partition overlap across all isoforms:")
display(pd.DataFrame(pairwise_overlap))

# Save the audit outputs.
cross_partition_compounds.to_csv(
    RESULTS_DIR / "cross_partition_compounds.csv",
    index=False
)

pd.DataFrame(pairwise_overlap).to_csv(
    RESULTS_DIR / "global_partition_overlap.csv",
    index=False
)

print("\nAudit outputs saved to:", RESULTS_DIR)
```

```python

from itertools import combinations
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

df = pd.read_csv(
    PROJECT_ROOT / "data" / "processed" / "cyp_substrates_cleaned.csv"
)

# For each canonical compound, list every isoform/partition combination.
compound_locations = (
    df.groupby("smiles_canonical")
      .apply(
          lambda g: sorted(set(
              zip(g["isoform"], g["split"])
          )),
          include_groups=False
      )
      .rename("locations")
      .reset_index()
)

# Count cross-partition sharing between different CYP isoforms.
records = []

for _, row in compound_locations.iterrows():
    locations = row["locations"]

    for loc1, loc2 in combinations(locations, 2):
        iso1, split1 = loc1
        iso2, split2 = loc2

        if iso1 != iso2 and split1 != split2:
            records.append({
                "isoform_1": iso1,
                "split_1": split1,
                "isoform_2": iso2,
                "split_2": split2,
                "smiles_canonical": row["smiles_canonical"]
            })

cross_isoform_leakage = pd.DataFrame(records)

if len(cross_isoform_leakage):
    summary = (
        cross_isoform_leakage
        .groupby([
            "isoform_1", "split_1",
            "isoform_2", "split_2"
        ])["smiles_canonical"]
        .nunique()
        .reset_index(name="shared_compounds")
        .sort_values("shared_compounds", ascending=False)
    )
    display(summary)
else:
    print("No cross-isoform cross-partition overlaps detected.")

out_dir = PROJECT_ROOT / "results" / "metrics"
out_dir.mkdir(parents=True, exist_ok=True)

cross_isoform_leakage.to_csv(
    out_dir / "cross_isoform_partition_overlap_details.csv",
    index=False
)

print("Detailed overlap records:", len(cross_isoform_leakage))
```
